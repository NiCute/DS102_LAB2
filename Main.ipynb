- Use Numpy only to construct the Logistic Regression model.
- Train and evaluating (precision, recall, f1) the Logistic Regression model using the Gradient Descent approach **to classify 0 and 1 digit images** on the [MNIST](https://github.com/cvdfoundation/mnist?tab=readme-ov-file) dataset.
- Visualize the loss function of the training process.

```python
images = idx2numpy.convert_from_file('train-images.idx3-ubyte')
labels = idx2numpy.convert_from_file('train-labels.idx1-ubyte')
```

```python
for i in range(4):
    plt.imshow(images[i])
    plt.title(labels[i])
    plt.show()
```

```python
def doc_anh(path):
    data = np.fromfile(path, dtype = np.uint8)
    data = data[16:].reshape(-1, 784)
    return data / 255

def doc_nhan(path):
    labels = np.fromfile(path, dtype = np.uint8)
    return labels[8:]
```

```python
x_train = doc_anh('train-images.idx3-ubyte')
y_train = doc_nhan('train-labels.idx1-ubyte')
x_test = doc_anh('t10k-images.idx3-ubyte')
y_test = doc_nhan('t10k-labels.idx1-ubyte')

x_train = [x_train[i] for i in range(len(y_train)) if y_train[i] == 0 or y_train[i] == 1]
y_train = [y_train[i] for i in range(len(y_train)) if y_train[i] == 0 or y_train[i] == 1]
x_train = np.array(x_train)
y_train = np.array(y_train)

model = Logistic_Regression_NumPy()
loss = model.Gradient_Descent_Logistic(x_train, y_train, 0.01, 1000)

y_pred = model.predict(x_test)
ketqua = model.evaluate(y_test, y_pred)
print("Ket qua danh gia la ", ketqua)

plt.plot(loss)
plt.title("Loss function")
plt.xlabel("Loops")
plt.ylabel("Loss")
plt.show()
```

- Use Numpy only to construct the Softmax Regression model.
- Train and evaluating (precision, recall, f1) the Softmax Regression model using the Gradient Descent approach **to classify 10 digit images** on the [MNIST](https://github.com/cvdfoundation/mnist?tab=readme-ov-file) dataset.
- Visualize the loss function of the training process.

```python
from softmax_regression import Softmax_Regression_Numpy

x_train = doc_anh('train-images.idx3-ubyte')
y_train = doc_nhan('train-labels.idx1-ubyte')
x_test = doc_anh('t10k-images.idx3-ubyte')
y_test = doc_nhan('t10k-labels.idx1-ubyte')

model_sm = Softmax_Regression_Numpy()
loss_sm = model_sm.Gradient_Descent_SoftMax(x_train, y_train, 0.1, 100)

y_pred_sm = model_sm.predict(x_test)
ketqua_sm = model_sm.evaluate(y_test, y_pred_sm)
print("Ket qua danh gia la ", ketqua_sm)

plt.plot(loss_sm)
plt.title("Loss function")
plt.xlabel("Loops")
plt.ylabel("Loss")
plt.show()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import precision_score, recall_score, f1_score

#logistic regression
x_train = doc_anh('train-images.idx3-ubyte')
y_train = doc_nhan('train-labels.idx1-ubyte')
x_test = doc_anh('t10k-images.idx3-ubyte')
y_test = doc_nhan('t10k-labels.idx1-ubyte')

x_train = [x_train[i] for i in range(len(y_train)) if y_train[i] == 0 or y_train[i] == 1]
y_train = [y_train[i] for i in range(len(y_train)) if y_train[i] == 0 or y_train[i] == 1]
x_test = [x_test[i] for i in range(len(y_test)) if y_test[i] == 0 or y_test[i] == 1]
y_test = [y_test[i] for i in range(len(y_test)) if y_test[i] == 0 or y_test[i] == 1]
x_train = np.array(x_train)
y_train = np.array(y_train)

model1 = LogisticRegression()
model1.fit(x_train, y_train)
y_pred1 = model1.predict(x_test)

print('Ket qua danh gia logistic la:')
print('precision:', precision_score(y_test, y_pred1))
print('recall:', recall_score(y_test, y_pred1))
print('f1_score:', f1_score(y_test, y_pred1))

x_train = doc_anh('train-images.idx3-ubyte')
y_train = doc_nhan('train-labels.idx1-ubyte')
x_test = doc_anh('t10k-images.idx3-ubyte')
y_test = doc_nhan('t10k-labels.idx1-ubyte')

model2 = LogisticRegression()
model2.fit(x_train, y_train)
y_pred2 = model2.predict(x_test)

print('Ket qua danh gia softmax la:')
print('precision:', precision_score(y_test, y_pred2, average = 'macro'))
print('recall:', recall_score(y_test, y_pred2, average = 'macro'))
print('f1_score:', f1_score(y_test, y_pred2, average = 'macro'))
```
